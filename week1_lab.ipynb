```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
url = "https://raw.githubusercontent.com/mwaskom/seaborn-data/master/penguins.csv"
penguins = pd.read_csv(url)
```

```python
penguins.head()
```

Which penguin species has the greatest average
body mass? Calculate the mean body mass for each species:

```python
penguins.groupby("species")["body_mass_g"].mean()
```

```python
plt.scatter(
    penguins["bill_length_mm"],
    penguins["flipper_length_mm"]
)

plt.xlabel("Bill lenght (mm)")
plt.ylabel("flipper lenght (mm)")
plt.title("Penguin Bill lenght vs. Flipper lenght")

plt.show()
```
